```python
import pandas as pd # Loads the Pandas library, which is used for handling data tables
import numpy as np # Loads the NumPy lirary, which handles high performance mathematical operations and arrays
import matplotlib.pyplot as plt # Loads the Pyplot interface from Matplotlib, which is sed to create graphs and charts

print("Libraries successfully set up.")
```

```python
# Load and inspect dataset
df_raw = pd.read_csv("/content/drive/MyDrive/CariSurg MedTech Pathways 2026/Week 0/Data/EmergencyTriageDataset_Reduced_Dirty.csv")
print(f"Dataset loaded: {df_raw.shape[0]} rows x {df_raw.shape[1]} columns")
print(f"\nColumns: {list(df_raw.columns)}\n")

df_raw.head()
```

```python
# Inspect gender column
print("Unique values in Gender:")
print(df_raw['Gender'].unique())
print(f"\nTotal unique values: {df_raw['Gender'].nunique()}")
print("\nGender value counts:")
print(df_raw['Gender'].value_counts())
```

```python
# Build mapping for Gender
gender_map = {
    '1': 1,
    'MALE':1,
    'Male':1,
    'FEMALE':0,
    'Female':0,
    '0':0,
}

# Create new column for cleaned gender data
df_raw['Gender Clean'] = df_raw['Gender'].map(gender_map)
```

```python
# Verify cleaning
print("After cleaning:")
print(df_raw['Gender Clean'].value_counts())
print(f"\nAny NaN values? {df_raw['Gender'].isnull().sum()}")
```

```python
# Drop the original dirty column
df_raw = df_raw.drop(columns=['Gender'])
df_raw = df_raw.rename(columns={'Gender Clean': 'Gender'})

# Confirming result
print("Column 'Gender' after cleaning:")
print(df_raw['Gender'].value_counts())
print(f"Data type: {df_raw['Gender'].dtype}")
print(f"\nAny NaN values? {df_raw['Gender'].isnull().sum()}\n")
df_raw.head()
```

```python
# Create new cvs file with cleaned Gender column
df_raw.to_csv("/content/drive/MyDrive/CariSurg MedTech Pathways 2026/Week 0/Data/EmergencyTriageDataset_Reduced_Gender_Cleaned.csv", index=False)
```

📓 Lessons Learnt


- I found that splitting my work into clear steps (loading, inspecting, cleaning, verifying) made my code easier to understand and debug.

- I learnt to always check outputs using .head(), .unique(), and .value_counts() after each change. This helped me confirm everything was working correctly before moving on.

- I realised I should avoid overwriting original columns too early. Keeping a clean version alongside the raw data makes the workflow safer and more flexible.

- When I deleted the 'Gender' column and renamed 'Gender Clean' --> 'Gender', the notebook broke on re-run because the 'Gender' column no longer existed. This showed me that my workflow wasn’t fully  reproducible and idempotent (new word to my vocab, haha!). In future, I would either:
   - keep both the raw and cleaned columns, or
   - maintain an untouched copy of the original dataset and only transform a working copy.
